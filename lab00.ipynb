```python
#pandas is a principal lib of data analysis with python.
import pandas as pd 
```

### Dataframe is a table

```python
dic = {
    'Yes': [50, 21],
    'No': [131, 2]
}
pd.DataFrame(dic)
```

```python
dic_with_string = {
    "Nike": ["Running shoes", "Football shoes", "Tennis shoes"],
    "Adidas": ["Running shoes", "Football shoes", "Tennis shoes"]
}
pd.DataFrame(dic_with_string)

dic_with_index = {
    "Nike": ["Running shoes", "Football shoes", "Tennis shoes"],
    "Adidas": ["Running shoes", "Football shoes", "Tennis shoes"]
}
pd.DataFrame(dic_with_index, index=["Model 1", "Model 2", "Model 3"])
```

### Series is a list

```python
list_a = [1, 2, 3, 4, 5]
pd.Series(list_a)
```

```python
pd.Series(list_a, index=["a", "b", "c", "d", "e"], name="My Series")
```
