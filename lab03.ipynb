### 2 - O que poder estar errado?
- Faltantes
- Duplicatas
- Tipos

```python
df.isnull().sum() # show the total number of null values in each column
```

```python
# exemplo: preencher com a mediana (robusta a extremos)
m = df['price'].median()
df['price'] = df['price'].fillna(m)
```

```python
df[df.duplicated(keep='first')] # identify duplicate rows in the dataset
df.drop_duplicates(keep='first', inplace=True)  # remove if any duplicate rows in the dataset
```

### Fase 3 — Visualização e relações

```python
fig = sns.histplot(data=df, x='price', kde=True, color='b')
fig.set_title('Distribuição da Variável Price', size=18)
fig.set_xlabel('Price', size=14); fig.set_ylabel('Count', size=14)
```

```python
df['brand'] = df['CarName'].str.split(' ').str[0]
sns.countplot(data=df, y='brand')
```

```python
df['brand'].unique()           # descobrir os erros
 
df['brand'] = df['brand'].replace({
    'maxda': 'mazda', 'toyouta': 'toyota',
    'vokswagen': 'volkswagen', 'vw': 'volkswagen',
    'porcshce': 'porsche'})

sns.boxplot(data=df, y='price')
```
