```python
# Preparação do ambiente

# Importação de bibliotecas
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn import datasets, linear_model
from sklearn.model_selection import train_test_split     # Train/Test
from sklearn.model_selection import KFold                 # Cross-Validation
```

```python
# Leitura do dataset Diabetes, do sklearn
columns = "age sex bmi bp tc ldl hdl tch ltg glu".split()
diabetes = datasets.load_diabetes()
df = pd.DataFrame(diabetes.data, columns=columns)
y = diabetes.target         # alvo (variável dependente da regressão)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df, y, test_size=0.2, random_state=42)      # random_state p/ reprodutibilidade
 
print(X_train.shape, y_train.shape)
print(X_test.shape,  y_test.shape)
```
